# Ouroboros Google Colab Quickstart

Welcome to Ouroboros! This notebook runs full source-mode Ouroboros inside Google Colab headlessly (without local OS windows) and initializes the Telegram Control Bridge automatically so you can chat directly with your agent.

### 🔑 Prerequisites & Setup
Ouroboros requires a model provider API key (e.g., OpenRouter, OpenAI, or Anthropic API keys) and a Telegram Bot Token.

We recommend configuring these secrets securely in Google Colab:
1. **Google Colab Secrets Manager (🔑 Sidebar Menu)**:
   - Click on the **Key icon (🔑)** in the left sidebar menu of Colab.
   - Add the following secrets and **toggle/grant notebook access** for each:
     - `TELEGRAM_BOT_TOKEN`
     - `OPENROUTER_API_KEY` (Our default multi-model aggregator)
     - `OPENAI_API_KEY` (Optional)
     - `ANTHROPIC_API_KEY` (Optional)
     - `GITHUB_TOKEN` (Optional, to push updates back to your personal origin repository)
2. **Interactive Prompts (Fallback)**:
   - If you do not configure them in the sidebar, running the code cell below will securely prompt you for these values (using your browser's hidden password prompt) so they are never hardcoded or saved in the notebook.

### 💾 Persistent State on Google Drive
This quickstart automatically mounts your Google Drive at `/content/drive`. Ouroboros will create a folder at `MyDrive/Ouroboros` to securely persist your configuration, settings, agent memories (`identity.md`, `scratchpad.md`), and custom skills so they survive ephemeral Colab session restarts.
